```python
from app.llmops.factory import get_chat_model
```

```python
gemini_llm = get_chat_model(provider="gemini")
response = gemini_llm.invoke("Chào bạn, bạn là ai?")
print(response.text)
```

```python
local_llm = get_chat_model(provider="openai_compat")
response = local_llm.invoke("Chào bạn, bạn là ai?")
print(response.content)
```

```python
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage

@tool
def search_docs(query: str) -> str:
    """Tìm tài liệu nội bộ liên quan tới query."""
    return f"kết quả tìm kiếm cho: {query}"

llm_with_tools = get_chat_model(provider="openai_compat").bind_tools([search_docs])
ai_msg = llm_with_tools.invoke([HumanMessage(content="Tìm giúp tôi tài liệu về chính sách vay vốn")])

print("tool_calls:", ai_msg.tool_calls)
print("content:", ai_msg.content)
```
