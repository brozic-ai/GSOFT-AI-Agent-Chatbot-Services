```python
import os
from pathlib import Path
from dotenv import load_dotenv
from google import genai
from google.genai import types

load_dotenv()

client = genai.Client(
    api_key=os.getenv("GEMINI_API_KEY")
)

model_name = os.getenv("GEMINI_MODEL")

PROJECT_ROOT = Path.cwd().parent

PDF_PATH = (
    PROJECT_ROOT
    / "storage"
    / "projects"
    / "proj-namabank-01"
    / "documents"
    / "contracts"
    / "nam_a_bank.pdf"
)

print(f"PDF exists: {PDF_PATH.exists()}")
print(f"PDF path: {PDF_PATH}")
```

```python
uploaded_file = client.files.upload(
    file=PDF_PATH
)

print(uploaded_file)
```

```python
response = client.models.generate_content(
    model=model_name,

    contents=[
        uploaded_file,
        """
        Hãy đọc file hợp đồng PDF tiếng việt này cho tôi.

        Yêu cầu:
        - Trả về nội dung: 
            + khách hàng: [đối tượng khách hàng trong file]
            + Nội dung: [tóm tắt nội dung yêu cầu phần mềm của hợp đồng dưới 100 chữ và không được để số tiền]
        """
    ]
)

text = response.text

print(text)
```
